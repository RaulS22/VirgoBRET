```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE
```

Here is important that we use the same half width window. The chosen one was 12s.

```python
RANDOM_STATE = 42
METRIC = 'euclidean'
PERPLEXITY = 350
```

```python
df_ont20_2022_hw1 = pd.read_parquet("year_TERZ_parquets/TERZ_HHZ_hw1_ont20_2022.parquet")
df_ont20_2022_hw2 = pd.read_parquet("year_TERZ_parquets/TERZ_HHZ_hw2_ont20_2022.parquet")
df_ont20_2022_hw4 = pd.read_parquet("year_TERZ_parquets/TERZ_HHZ_hw4_ont20_2022.parquet")
df_ont20_2022_hw6 = pd.read_parquet("year_TERZ_parquets/TERZ_HHZ_hw6_ont20_2022.parquet")
df_ont20_2022_hw12 = pd.read_parquet("year_TERZ_parquets/TERZ_HHZ_hw12_ont20_2022.parquet")
```

```python
print(f"Number of triggers 2022: hw1={df_ont20_2022_hw1.shape[0]}, hw2={df_ont20_2022_hw2.shape[0]}, hw4={df_ont20_2022_hw4.shape[0]}, hw6={df_ont20_2022_hw6.shape[0]}, hw12={df_ont20_2022_hw12.shape[0]}")
```

We will compare ont=15, ont=20 and ont=25 for the same year

2022 hw=1s

```python
df = df_ont20_2022_hw1.copy()

feature_columns = [col for col in df.columns if col.startswith("feature_")]
if len(feature_columns) == 0:
    raise ValueError("No feature_* columns found.")

print(f"Number of features: {len(feature_columns)}")

X = df[feature_columns].to_numpy(dtype="float64")
print(f"X.shape = {X.shape}")
X_max = X.max(axis=1, keepdims=True)
```

```python
X_norm = X / X_max

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_norm)
print(f"Scaled matrix: {X_scaled.shape}") 
```

```python
tsne = TSNE(n_components=2,perplexity=PERPLEXITY,early_exaggeration=14,init="pca",learning_rate="auto",random_state=RANDOM_STATE, metric=METRIC)
X_tsne = tsne.fit_transform(X_scaled)
```

```python
fig, ax = plt.subplots(figsize=(12, 10))
scatter = ax.scatter(X_tsne[:, 0], X_tsne[:, 1],s=5,alpha=0.5)
ax.set_title(f"Perplexity = {PERPLEXITY} | Early Exaggeration = 14 | PCA Components = No PCA | Metric = {METRIC}")
ax.set_xlabel("t-SNE 1")
ax.set_ylabel("t-SNE 2")

plt.tight_layout()
plt.show()
```
