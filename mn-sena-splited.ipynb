```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
from pathlib import Path

import re
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE
import umap
```

```python
METRIC = "euclidean"
```

2022

```python
RANDOM_STATE = 42
df = pd.read_parquet("year_MNSENA_results_cut/qtransform_features_2022.parquet")

feature_columns = [col for col in df.columns if col.startswith("feature_")]
if len(feature_columns) == 0:
    raise ValueError("No feature_* columns found.")

print(f"Number of features: {len(feature_columns)}")

X = df[feature_columns].to_numpy(dtype="float64")
print(f"X.shape = {X.shape}")

X_max = X.max(axis=1, keepdims=True)
```

```python
X_norm = X / X_max

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_norm)
print(f"Scaled matrix: {X_scaled.shape}") 
```

```python
tsne = TSNE(n_components=2,perplexity=250,early_exaggeration=14,init="pca",learning_rate="auto",random_state=RANDOM_STATE, metric=METRIC)
X_tsne = tsne.fit_transform(X_scaled)
```

```python
fig, ax = plt.subplots(figsize=(12, 10))
scatter = ax.scatter(X_tsne[:, 0], X_tsne[:, 1],s=5,alpha=0.5)
ax.set_title(f"Perplexity = 250 | Early Exaggeration = 14 | PCA Components = No PCA | Metric = {METRIC}")
ax.set_xlabel("t-SNE 1")
ax.set_ylabel("t-SNE 2")

plt.tight_layout()
plt.show()
```

```python
reducer = umap.UMAP() 

embedded = reducer.fit_transform(X_scaled)
embedded.shape
```

```python
fig, ax = plt.subplots(figsize=(12, 10))
scatter = ax.scatter(embedded[:, 0], embedded[:, 1],s=5,alpha=0.5)
ax.set_title(f"UMAP")
ax.set_xlabel("UMAP 1")
ax.set_ylabel("UMAP 2")

plt.tight_layout()
plt.show()
```
